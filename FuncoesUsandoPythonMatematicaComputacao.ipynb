```python
#################################################################
## Matemática para Computação - Relações e Funções
#################################################################
## EXEMPLO 4
#################################################################
import numpy as np
import matplotlib.pyplot as plt

def P(x):
    return 2 * x - 3

def zero_f(a0,a1):
    return -a0/a1

x = np.linspace(-10, 10, 1000)
y = P(x)
a1 = 2
a0 = -3
x_zero = zero_f(a0,a1)
y_zero = P(x_zero)

plt.figure(figsize=(8, 5))
plt.plot(x, y, color = 'blue')

plt.plot(x_zero, y_zero, 'ro')

plt.annotate(
    '(%.1f, %.1f)' % (x_zero, y_zero),
    xy = (x_zero, y_zero),
    xytext = (x_zero - 1, y_zero - 3.5),
    fontsize = 10,
    color = 'red'
)
plt.axhline(0, color = 'black', linewidth = 0.8)
plt.axvline(0, color = 'black', linewidth = 0.8)

plt.title('Gráfico da função P(x) = 2x - 3')
plt.xlabel('x')
plt.ylabel('P(x)')
plt.grid()
plt.show()
```

```python
#################################################################
## EXERCÍCIO 3
#################################################################
import numpy as np
import matplotlib.pyplot as plt

def P(x):
    return -x**2 + x + 2

def encontrar_zeros(a, b, c):
    delta = b**2 - 4*a*c
    if delta < 0:
        return []
    elif delta == 0:
        x = -b / (2*a)
        return [x]
    else:
        raiz_delta = (delta)**(0.5)
        x1 = (-b + raiz_delta) / (2*a)
        x2 = (-b - raiz_delta) / (2*a)
        return [x1, x2]

a = -1
b = 1
c = 2

zeros = encontrar_zeros(a, b, c)

x = np.linspace(-10, 10, 400)
y = P(x)

plt.figure(figsize=(8, 5))
plt.plot(x, y, label = 'P(x) = x^2 - 3x + 2', color = 'blue')

for x_zero in zeros:
    y_zero = P(x_zero)
    plt.plot(x_zero, y_zero, 'ro')
    plt.annotate(
        '(%.1f, %.1f)' % (x_zero, y_zero),
        xy = (x_zero, y_zero),
        xytext = (x_zero - 0.2, y_zero + 3),
        fontsize = 10,
        color = 'red'
    )

plt.axhline(0, color = 'black', linewidth = 0.8)
plt.axvline(0, color = 'black', linewidth = 0.8)

# Personalizações
plt.title('Gráfico da função P(x) = - x^2 + x + 2')
plt.xlabel('x')
plt.ylabel('P(x)')
plt.grid()
plt.show()
```

```python
#################################################################
## EXEMPLO 6
#################################################################
import numpy as np
import matplotlib.pyplot as plt

def R(x):
    return 1 / (x**2 - 1)

x1 = np.linspace(-5, -1.01, 500)
x2 = np.linspace(-0.99, 0.99, 500)
x3 = np.linspace(1.01, 5, 500)

y1 = R(x1)
y2 = R(x2)
y3 = R(x3)

print('R(%.1f) = %.1f' % (0,R(0)))
print('R(%.1f) = %.1f' % (2,R(2)))

plt.figure(figsize=(10, 6))
plt.plot(x1, y1, color = 'blue')
plt.plot(x2, y2, color = 'blue')
plt.plot(x3, y3, color = 'blue')

plt.axvline(x = -1, color = 'red', linestyle = '--')
plt.axvline(x = 1,  color = 'red', linestyle = '--')

plt.title(r'Gráfico da função $R(x) = \frac{1}{x^2 - 1}$')
plt.xlabel('x')
plt.ylabel('R(x)')
plt.ylim(-10, 10)
plt.grid()
plt.show()
```

```python
#################################################################
## EXEMPLO 7
#################################################################
import numpy as np
import matplotlib.pyplot as plt
import math

def f(x):
    return np.log10(x)

x_raiz = 1
y_raiz = f(x_raiz)

x1 = 10
x2 = 100

print('f(%.1f) = %.1f' % (x1,f(x1)))
print('f(%.1f) = %.1f' % (x2,f(x2)))

x_vals = np.linspace(0.1, 150, 400)
y_vals = f(x_vals)

plt.figure(figsize=(9, 6))
plt.plot(x_vals, y_vals, color = 'magenta')

plt.plot(x_raiz, y_raiz, 'bo')

plt.annotate('(%.1f, %.1f)' % (x_raiz, y_raiz),
             xy = (x_raiz, y_raiz), xytext = (x_raiz + 0.6, y_raiz + 0.1),
             color = 'blue')

plt.axhline(0, color = 'black', linewidth = 0.8)
plt.axvline(0, color = 'black', linewidth = 0.8)

plt.title(r'Gráfico da função $f(x) = \log_{10}(x)$')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid()
plt.show()
```

```python
#################################################################
## EXEMPLO 8
#################################################################
import numpy as np
import matplotlib.pyplot as plt

def g(x):
    return 0.2**x

x1 = 0
x2 = 0.2

print('g(%.1f) = %.1f' % (x1, g(x1)))
print('g(%.1f) = %.1f' % (x2, g(x2)))

x_vals = np.linspace(-5, 2, 400)
y_vals = g(x_vals)

plt.figure(figsize=(9, 6))
plt.plot(x_vals, y_vals, color = '#AEC6CF')

plt.axhline(0, color = 'black', linewidth = 0.8)
plt.axvline(0, color = 'black', linewidth = 0.8)

plt.title(r'Gráfico da função $g(x) = 0.2^x$')
plt.xlabel('x')
plt.ylabel('g(x)')
plt.grid()
plt.show()
```

```python
#################################################################
## EXEMPLO 9
#################################################################
import numpy as np
import matplotlib.pyplot as plt

x_vals = np.linspace(-2 * np.pi, 2 * np.pi, 1000)
y_vals = np.cos(x_vals)
w_vals = np.sin(x_vals)
z_vals = np.tan(x_vals)

asymptotes_tan = [-(3*np.pi)/2, -np.pi/2, np.pi/2, (3*np.pi)/2]

tolerance = 0.1
for asymptote in asymptotes_tan:
    z_vals[(x_vals > asymptote - tolerance) & (x_vals < asymptote + tolerance)] = np.nan

plt.figure(figsize=(10, 6))
plt.plot(x_vals, y_vals, color = '#9bfab0', label='cos(x)')
plt.plot(x_vals, w_vals, color = '#AEC6CF', label='sin(x)')
plt.plot(x_vals, z_vals, color = 'red', label='tan(x)')

plt.axhline(0, color = 'black', linewidth = 0.8)
plt.axvline(0, color = 'black', linewidth = 0.8)

for asymptote in asymptotes_tan:
    plt.axvline(x = asymptote, color = 'red', linestyle = '--')

plt.title(r'Gráfico das funções $f(x) = \sin(x)$, $g(x) = \cos(x)$ e $h(x) = \tan(x)$')
plt.xlabel('x')
plt.ylabel('f(x), g(x), h(x)')
plt.ylim(-2, 2)
plt.grid()
plt.legend()
plt.show()
```
